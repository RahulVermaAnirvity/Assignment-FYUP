```python
import pandas as pd

# --- Load Data ---
file_name = 'fyup_advanced_scan_results.csv'
try:
    df = pd.read_csv(file_name)
    print(f"✅ Successfully loaded '{file_name}'.")
    print("----------------------------------------")
    
    # --- Basic Analysis ---
    total_matches = len(df)
    unique_universities_count = df['University'].nunique()
    keyword_counts = df['Keyword_Found'].value_counts()

    print(f"Total Keyword Matches Found: {total_matches}")
    print(f"Unique Universities with Matches: {unique_universities_count}")
    print("\n--- Keyword Frequency ---")
    print(keyword_counts)
    print("----------------------------------------")

    # --- Example Filtering ---
    print("\n--- Example: All results for 'www.ignou.ac.in' ---")
    ignou_results = df[df['University'] == 'www.ignou.ac.in']
    print(ignou_results[['Keyword_Found', 'Context']].head()) # Show first 5
    
except FileNotFoundError:
    print(f"❗️ ERROR: Make sure the file '{file_name}' is in the same folder.")
```
